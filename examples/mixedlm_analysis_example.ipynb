# Mixed-effects analysis from model-input data

```
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().parent if Path.cwd().name == "examples" else Path.cwd()
sys.path.insert(0, str(REPO_ROOT))

from mixedlm_analysis import *
```

```
input_path = REPO_ROOT / "model_input_data.xlsx"
outdir = REPO_ROOT / "mixedlm_outputs"
figdir = REPO_ROOT / "mixedlm_figures"

config = MixedLMConfig(
    y_col="creep_at_10s",
    stage_col="stage",
    individual_col="individual_id",
    probe_size_col="probe_size",
    x_position_col="x_position",
    y_position_col="y_position",
)

print(input_path)
print(input_path.exists())
```

```
analysis = run_mixedlm_analysis(
    input_path=input_path,
    outdir=outdir,
    config=config,
    save_outputs=True,
    prefix="mixedlm",
)

analysis.fixed_effects
```

```
make_standard_diagnostic_plots(
    analysis,
    outdir=figdir,
    config=config,
    show=False,
)

print(analysis.model_result.summary())
```
